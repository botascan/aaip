```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
```

```python
world = gpd.read_file("data/naturalearth_lowres/naturalearth_lowres.shp")
afrika = world[world["continent"] == "Afrika"].copy()
```

```python
df = pd.read_csv(
    "API_IT.NET.USER.ZS_DS2_en_csv_v2_62387/API_IT.NET.USER.ZS_DS2_en_csv_v2_62387.csv",
    skiprows=4
)
df["Country Code"]
```

```python
zimbabwe = df[df["Country Code"] == "ZWE"]
zimbabwe = zimbabwe.T.dropna()
zimbabwe.tail()
```

```python
zimbabwe.columns = ["internet_users_pct"]
years = zimbabwe.iloc[5:, :]
series = years
```

```python
fig, ax = plt.subplots(figsize=(9,5))
ax.plot(series.index, series["internet_users_pct"], marker="o")
ax.set_xlabel("Year")
ax.set_ylabel("Individuals using the Internet (% of population)")
ax.set_title("Internet usage in Zimbabwe over time")
ax.grid(True, alpha=0.3)
plt.xticks(rotation=90)
```
