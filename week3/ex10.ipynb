```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

```python
surv_rate = df.groupby("Pclass")["Survived"].mean()
surv_rate
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
bars = ax.bar(["1st", "2nd", "3rd"], surv_rate * 100,
              color=["#2a9d8f", "#e9c46a", "#e76f51"])
ax.bar_label(bars, fmt="%.1f%%")                     # show values on the bars
ax.set_title("Survival Rate by Ticket Class on the Titanic")
ax.set_xlabel("Ticket class")
ax.set_ylabel("Share of passengers who survived (%)")
ax.set_ylim(0, 100)
plt.tight_layout()
plt.show()
```
