```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
df = pd.read_csv("titanic.csv")

flows = (df.dropna(subset=["Embarked"])
         .groupby(["Embarked", "Survived"])
         .size()
         .reset_index(name="Count"))
flows
```

```python
port_names = {"C": "Cherbourg", "Q": "Queenstown", "S": "Southampton"}
outcomes = {0: "Died", 1: "Survived"}

labels = list(port_names.values()) + list(outcomes.values())

port_idx = {"C": 0, "Q": 1, "S": 2}
outcome_idx = {0: 3, 1: 4}

source = flows["Embarked"].map(port_idx)
target = flows["Survived"].map(outcome_idx)
value  = flows["Count"]
```

```python
node_colors = ["#264653", "#287271", "#8ab17d", "#e76f51", "#2a9d8f"]
# link color = color of the target (red for died, green for survived), made transparent
link_colors = ["rgba(231,111,81,0.45)" if t == 3 else "rgba(42,157,143,0.45)"
               for t in target]

fig = go.Figure(go.Sankey(
    node=dict(label=labels, color=node_colors, pad=20, thickness=20),
    link=dict(source=source, target=target, value=value, color=link_colors)
))

fig.update_layout(
    title_text="Passenger Flow: Port of Embarkation → Survival Outcome",
    font_size=13
)
#fig.show()

fig.add_annotation(x=0.0, y=1.08,
                   text="<b>Port of embarkation</b>", showarrow=False)
fig.add_annotation(x=1.0, y=1.08,
                   text="<b>Outcome</b>", showarrow=False)
```
